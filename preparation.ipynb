Разделение датасет на train, test, val 70/15/15

```python
import random

RANDOM_SEED = 22
random.seed(RANDOM_SEED)
```

```python
import os
import random
import shutil

images_dir = "number_car_detect"
labels_dir = "labels/obj_train_data"
output_dir = "dataset"

split_ratio = {
    "train": 0.7,
    "val": 0.15,
    "test": 0.15
}

for split in ["train", "val", "test"]:
    os.makedirs(f"{output_dir}/images/{split}", exist_ok=True)
    os.makedirs(f"{output_dir}/labels/{split}", exist_ok=True)

files = []
img_formats = [".png", ".jpg"]

for label_file in os.listdir(labels_dir):
    if label_file.endswith(".txt"):
        label_path = os.path.join(labels_dir, label_file)

        with open(label_path, "r") as f:
            lines = [line.strip() for line in f.readlines() if line.strip()]
            if len(lines) == 0:
                continue

        base_name = os.path.splitext(label_file)[0]

        for fmt in img_formats:
            img_file = base_name + fmt
            if os.path.exists(os.path.join(images_dir, img_file)):
                files.append(img_file)
                break

random.shuffle(files)

print("Всего файлов:", len(files))

n = len(files)
train_end = int(n * split_ratio["train"])
val_end = train_end + int(n * split_ratio["val"])

splits = {
    "train": files[:train_end],
    "val": files[train_end:val_end],
    "test": files[val_end:]
}

for split, split_files in splits.items():
    for img in split_files:
        base_name = os.path.splitext(img)[0]
        label = base_name + ".txt"

        shutil.copy(os.path.join(images_dir, img),
                    f"{output_dir}/images/{split}/{img}")

        shutil.copy(os.path.join(labels_dir, label),
                    f"{output_dir}/labels/{split}/{label}")

print("Готово!")
```
